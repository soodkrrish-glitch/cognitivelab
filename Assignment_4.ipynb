```python
import pandas as pd
roll_number = "1024170267"
```

```python
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

categories = ["billing", "account", "general"]
```

```python
last_two_digits = [int(digit) for digit in roll_number[-2:]]

personalized_entries = []

for digit in last_two_digits:
    category = categories[digit % 3]

    if category == "billing":
        entry = {
            "question": "how can i check my payment status",
            "answer": "You can check your payment status from the Billing section.",
            "keywords": "payment status billing transaction",
            "category": category
        }

    elif category == "account":
        entry = {
            "question": "how do i update my registered mobile number",
            "answer": "Go to Account Settings and update your registered mobile number.",
            "keywords": "mobile number update account",
            "category": category
        }

    else:
        entry = {
            "question": "how can i contact customer support",
            "answer": "You can contact customer support through the Help section.",
            "keywords": "support help contact assistance",
            "category": category
        }

    personalized_entries.append(entry)
```

```python
all_entries = fixed_entries + personalized_entries

df = pd.DataFrame(all_entries)

print("Q1: Final Knowledge Base")
print(df)
```

```python
def score_query(query, df):
    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        question_words = set(row["question"].lower().split())
        matched_words = query_words & (keywords | question_words)

        score = len(matched_words)

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results.sort(key=lambda x: x["score"], reverse=True)

    return results


query = input("\nEnter your query: ")

results = score_query(query, df)

print(" Matching Entries")

if results:
    for result in results:
        print(result)
else:
    print("No matching entries found.")
```

```python
def same_category(category_name, df):
    return df[df["category"].str.lower() == category_name.lower()]

selected_category = personalized_entries[0]["category"]

print("Entries in category:", selected_category)

print(same_category(selected_category, df)[["question", "category"]])
```

```python
print("Knowledge Base Entries")

print(df[["question", "keywords"]])

entry_index = int(input("Enter the index of the entry you want to update: "))
new_keyword = input("Enter a new keyword: ").strip().lower()

df.loc[entry_index, "keywords"] += " " + new_keyword

filename = roll_number + "_faq_data.csv"
df.to_csv(filename, index=False)

print("\nUpdated entry:")
print(df.loc[entry_index])

print("filename)
```

```python
def score_query_with_ties(query, df):
    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        question_words = set(row["question"].lower().split())

        matched_words = query_words & (keywords | question_words)

        score = len(matched_words)

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })
if not results:
        print("No matching entries found.")
        return

    highest_score = max(result["score"] for result in results)

    best_matches = [
        result for result in results
        if result["score"] == highest_score
    ]

    print("\nBest matching entries:")

    if len(best_matches) > 1:
        print("Tie detected! Multiple entries have the highest score.")

    for result in best_matches:
        print(result)

print("Tie Demonstration")

score_query_with_ties("fee", df)

print("Non-Tie Demonstration")

score_query_with_ties("password", df)
```
