```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(data)

employees.to_csv("employees.csv", index=False)

print(employees)
```

```python
print(employees.shape)
```

```python
print(employees.info())
```

```python
print(employees.describe())
```

```python
print(employees.head(5))
print(employees.tail(3))
```

```python
average_salary = employees["Salary"].mean()
print("Average Salary:", average_salary)
```

```python
total_bonus = employees["Bonus"].sum()
print("Total Bonus:", total_bonus)
```

```python
youngest_age = employees["Age"].min()
print("Youngest Age:", youngest_age)
```

```python
rating = employees["Age"].max()
print("Highest Rating:", rating)
```

```python
employees_sorted = employees.sort_values(by="Salary", ascending=False)

print(employees_sorted)
```

```python
def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance"] = employees["Rating"].apply(categorize_rating)

print(employees[["Name", "Rating", "Performance"]])
```

```python
print(employees.isnull().sum())
```

```python
employees.rename(columns={"Employee_ID": "ID"}, inplace=True)

print(employees)
```

```python
experienced_employees = employees[
    employees["Years_of_Experience"] > 5
]

print(experienced_employees)
```

```python
IT_employees = employees[
    employees["Department"] == "IT"
]

print(IT_employees)
```

```python
def calc_tax(sal):
    return 0.1*sal

employees["TAX"] = employees["Salary"].apply(calc_tax)

print(employees[["Name", "Salary", "TAX"]])
```

```python
employees.to_csv("modified_employees.csv", index=False)

print("Modified DataFrame saved successfully.")
```
