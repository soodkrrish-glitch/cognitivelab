```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr)
print(arr + 2)
print(arr * 3)
print(arr / 2)
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("Most frequent value:",most_frequent)
print("Indices:",indices)
```

```python
import numpy as np

Divyansh = np.linspace(10, 100, 25)

print("Array:")
print(Divyansh)

#dimensions
print("Dimensions:", Divyansh.ndim)

#shape
print("Shape:", Divyansh.shape)

#number of elements
print("Total elements:", Divyansh.size)

# data type
print("Data type:", Divyansh.dtype)

#number of bytes
print("Total bytes:", Divyansh.nbytes)

#reshape
transpose_array = Divyansh.reshape(25, 1)

print("Transpose using reshape():")
print(transpose_array)

#T attribute
print("Transpose using T:")
print(Divyansh.reshape(1, 25).T)
```

```python
import numpy as np

ucs420_Divyansh = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Divyansh)

print("Mean:", np.mean(ucs420_Divyansh))

print("Median:", np.median(ucs420_Divyansh))

print("Maximum:", np.max(ucs420_Divyansh))

print("Minimum:", np.min(ucs420_Divyansh))

print("Unique Elements:", np.unique(ucs420_Divyansh))

reshaped_ucs420_Divyansh = ucs420_Divyansh.reshape(4, 3)

print("Reshaped Array:")
print(reshaped_ucs420_Divyansh)

resized_ucs420_Divyansh = np.resize(ucs420_Divyansh, (2, 3))

print("Resized Array:")
print(resized_ucs420_Divyansh)
```
