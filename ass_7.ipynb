```python
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Original paragraph (5-6 sentences)
text = """
Artificial intelligence is transforming the world through innovative technology.
Machine learning helps computers learn from data and solve complex problems.
I enjoy exploring artificial intelligence because it makes technology more useful.
Developers create intelligent applications to improve healthcare and education.
Learning new technologies helps students develop better problem-solving skills.
"""

print("Original Text:")
print(text)

# 1. Convert to lowercase and remove punctuation using re
clean_text = re.sub(r'[^\w\s]', '', text.lower())

print("\nLowercase Text Without Punctuation:")
print(clean_text)

# 2. Tokenize into words and sentences
words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("\nWord Tokens:")
print(words)

print("\nSentence Tokens:")
print(sentences)

# 3. Compare split() and word_tokenize()
split_words = clean_text.split()
nltk_words = word_tokenize(clean_text)

print("\nUsing split():")
print(split_words)

print("\nUsing word_tokenize():")
print(nltk_words)

print("\nAre both results equal?", split_words == nltk_words)

# 4. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in nltk_words
    if word.isalpha() and word not in stop_words
]

print("\nWords After Stopword Removal:")
print(filtered_words)

# 5. Display word frequency distribution
freq_dist = FreqDist(filtered_words)

print("\nWord Frequency Distribution:")
print(freq_dist)

print("\nFrequency of Each Word:")
for word, frequency in freq_dist.items():
    print(word, ":", frequency)
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')

text = """
Artificial intelligence is transforming the world through innovative technology.
Machine learning helps computers learn from data and solve complex problems.
I enjoy exploring artificial intelligence because it makes technology more useful.
Developers create intelligent applications to improve healthcare and education.
Learning new technologies helps students develop better problem-solving skills.
"""

# 1. Extract words containing only alphabets using re.findall()
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

print("Alphabetic Words:")
print(words)

# 2. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word not in stop_words
]

print("\nWords After Stopword Removal:")
print(filtered_words)

# 3. Perform stemming using PorterStemmer
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("\nStemmed Words:")
print(stemmed_words)

# 4. Perform lemmatization using WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\nLemmatized Words:")
print(lemmatized_words)

# 5. Compare stemming and lemmatization
print("\nComparison:")
for word, stem, lemma in zip(
    filtered_words, stemmed_words, lemmatized_words
):
    print(word, "-> Stem:", stem, "| Lemma:", lemma)
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# 1. Create three short texts
texts = [
    "Artificial intelligence helps computers learn and solve problems.",
    "Machine learning uses data to build intelligent systems.",
    "Artificial intelligence and machine learning improve technology."
]

# 2. Bag of Words using CountVectorizer
count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)

print("Bag of Words Representation:")
print("Features:", count_vectorizer.get_feature_names_out())
print(bow_matrix.toarray())

# 3. TF-IDF using TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

features = tfidf_vectorizer.get_feature_names_out()
scores = tfidf_matrix.toarray()

print("\nTF-IDF Representation:")
print("Features:", features)
print(scores)

# 4. Print top 3 keywords from each text
print("\nTop 3 Keywords from Each Text:")

for i in range(len(texts)):
    top_indices = scores[i].argsort()[::-1][:3]

    print(f"\nText {i + 1}:")
    for index in top_indices:
        print(features[index], ":", round(scores[i][index], 4))
```

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Create two short texts
text1 = """
Artificial intelligence enables computers to learn from data.
AI helps automate tasks and solve complex problems.
Machine learning is an important part of artificial intelligence.
"""

text2 = """
Blockchain is a decentralized technology for storing data.
It provides secure transactions and transparent records.
Blockchain is widely used in cryptocurrency systems.
"""

# 2. Preprocess and tokenize both texts
def preprocess(text):
    text = text.lower()
    words = re.findall(r'\b[a-zA-Z]+\b', text)
    return words

words1 = preprocess(text1)
words2 = preprocess(text2)

print("Tokens in Text 1:")
print(words1)

print("\nTokens in Text 2:")
print(words2)

# 3(a). Calculate Jaccard Similarity using sets
set1 = set(words1)
set2 = set(words2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard = len(intersection) / len(union)

print("\nCommon Words:")
print(intersection)

print("\nJaccard Similarity:", round(jaccard, 4))

# 3(b). Calculate Cosine Similarity using TF-IDF
texts = [text1, text2]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(texts)

cosine = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)[0][0]

print("Cosine Similarity:", round(cosine, 4))

# 3(c). Analyze the results
print("\nAnalysis:")

if jaccard > cosine:
    print("Jaccard Similarity is higher than Cosine Similarity.")
elif cosine > jaccard:
    print("Cosine Similarity is higher than Jaccard Similarity.")
else:
    print("Both similarity scores are equal.")

print("Jaccard measures overlap between unique words.")
print("Cosine measures similarity between TF-IDF vectors.")
```

```python
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# 1. Write product reviews
reviews = [
    "This laptop is excellent and very fast. I love its performance.",
    "The phone has a beautiful design and an amazing camera.",
    "The product is terrible and the battery life is very poor.",
    "I am disappointed with the poor quality and slow performance.",
    "The product is okay. It works as expected."
]

# 2. Calculate polarity and subjectivity
positive_reviews = []

print("Sentiment Analysis Results:\n")

for i, review in enumerate(reviews, start=1):
    blob = TextBlob(review)

    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    # 3. Classify reviews
    if polarity > 0:
        sentiment = "Positive"
        positive_reviews.append(review)
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("Review:", i)
    print("Text:", review)
    print("Polarity:", round(polarity, 3))
    print("Subjectivity:", round(subjectivity, 3))
    print("Sentiment:", sentiment)
    print()

# 4. Generate a word cloud using positive reviews
positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(positive_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Positive Reviews")
plt.show()
```
