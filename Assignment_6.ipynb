<h4>Q2</h4>

```python
steps = np.array([ 
[5000, 6200, 7100], 
[8000, 7500, 9000], 
[4500, 5100, 4800], 
[9000, 8500, 9500]])
```

```python
print("Total steps:",np.sum(steps))
```

```python
print("Avg steps:",np.mean(steps))
```

```python
print("Max steps:",np.max(steps))
print("Min steps:",np.min(steps))
```

```python
daily_total = np.sum(steps, axis=0)
print("Total steps for each day:", daily_total)
```

```python
user_total = np.sum(steps, axis=1)
print("Total steps for each user:", user_total)
```

```python
position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum value:", position)
print("Maximum steps:", steps[position])
```

<h4>Q4</h4>

```python
X = np.array([ 
[6, 70, 3], 
[5, 50, 6], 
[8, 80, 2], 
[4, 30, 8] 
])
```

```python
y = np.array([40, 65, 30, 85]) 
```

```python
print(X.ndim)
```

```python
print(X.shape)
```

```python
print(X.T)
```

```python
print(X.T @ X )
```

```python
XtX = X.T @ X
inverse = np.linalg.inv(XtX)
print("X.T @ X:")
print(XtX)
print("Inverse of X.T @ X:")
print(inverse)
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("Coefficients:")
print(beta)
```

```python
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
